Dataset Name: Fake Reviews Dataset
Source: Kaggle (mexwell)
Link: https://www.kaggle.com/datasets/mexwell/fake-reviews-dataset
Description:
This dataset contains 40,000 product reviews (balanced 50:50).

20,000 CG → Computer-Generated (Fake) reviews
20,000 OR → Original (Real/Authentic) human-written reviews

Columns: Typically includes the review text and a label (CG or OR).
Purpose: Suitable for training and evaluating models for Fake Review Detection.Citation:
Salminen et al. (2022). Creating and detecting fake reviews of online products. Journal of Retailing and Consumer Services.


Column	Meaning	Use for Project?
deceptive	Whether the review is Fake or Truthful	✅ Yes (Target Label)
hotel	Hotel name	❌ No
polarity	Positive or Negative sentiment	❌ Optional
source	Where the review came from	❌ No
text	Review text	✅ Yes (Input Feature)

Deceptive Opinion Spam Corpus, which is one of the best datasets for fake review detection
